```python
# Cargar modelo (ejecutar una sola vez)
from beto_predict import predict
```

```python
# Edita esta lista y re-ejecuta (Ctrl+Enter) para probar
casos = [
    "Construcción de puesto de salud en la municipalidad de Cajamarca",
    "Ampliación del canal de riego para la municipalidad de Tacna",
    "Nuevo hospital en la municipalidad de Cajamarca",
]

for r in predict(casos, top_k=3):
    print(f"\n{r['texto']}")
    for label, score in r["predicciones"]:
        print(f"  {label:40s} {score*100:6.2f}%")
```
